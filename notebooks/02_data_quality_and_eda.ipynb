# 02 · Data quality and EDA

> Educational decision-support prototype trained on synthetic PaySim data. Outputs are risk scores and review priorities that help human investigators decide what to review first. This system makes no fraud or AML determination and performs no automatic blocking, account closure, customer risk rating, or regulatory reporting. Results on synthetic data do not establish real-world detection effectiveness, fairness, or regulatory suitability.

Displays artifacts produced by `python -m aml_triage profile` and `python -m aml_triage eda`. Human observations live in `reports/data_quality_narrative.md` and `reports/eda_narrative.md`. No logic is defined here; all parameters come from `configs/`.

```python
from pathlib import Path

import json
import pandas as pd
import yaml
from IPython.display import Image, Markdown, display

from aml_triage.config import load

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
cfg = load(ROOT / "configs" / "base.yaml")
REPORTS = ROOT / cfg.paths.reports_dir
PROCESSED = ROOT / cfg.paths.processed_dir
print("config hash:", cfg.config_hash())
```

## Data quality report

```python
display(Markdown((REPORTS / "data_quality.md").read_text()))
```

## Split manifest

```python
m = json.loads((PROCESSED / "split_manifest.json").read_text()) if (PROCESSED / "split_manifest.json").exists() else {}
display(pd.DataFrame({k: m.get(k, {}) for k in ["rows", "positives", "step_ranges"]}) if m else Markdown("_Run `python -m aml_triage split` first._"))
```
