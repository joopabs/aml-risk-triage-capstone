# 04 · Feature selection and PCA

> Educational decision-support prototype trained on synthetic PaySim data. Outputs are risk scores and review priorities that help human investigators decide what to review first. This system makes no fraud or AML determination and performs no automatic blocking, account closure, customer risk rating, or regulatory reporting. Results on synthetic data do not establish real-world detection effectiveness, fairness, or regulatory suitability.

Displays the outputs of `python -m aml_triage select-features` and `python -m aml_triage pca`. Both selectors and the PCA were fitted on the training split only (fit-scope records shown below). Human justifications live in `reports/feature_selection_narrative.md` and `reports/pca_narrative.md`.

```python
from pathlib import Path

import json
import pandas as pd
from IPython.display import Image, Markdown, display

from aml_triage.config import load

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
cfg = load(ROOT / "configs" / "base.yaml")
REPORTS = ROOT / cfg.paths.reports_dir
PROCESSED = ROOT / cfg.paths.processed_dir
print("config hash:", cfg.config_hash())
```

## Fit scope (must be train only)

```python
sel = json.loads((REPORTS / "feature_selection.json").read_text())
pca = json.loads((REPORTS / "pca_report.json").read_text())
display(pd.DataFrame({"MI selector": sel["fit_scope"]["mi"], "L1 selector": sel["fit_scope"]["l1"], "PCA": pca["fit_scope"]}).T)
```

## Feature selection report

```python
display(Markdown((REPORTS / "feature_selection.md").read_text()))
```

## PCA report

```python
display(Markdown((REPORTS / "pca_report.md").read_text().replace("](figures/", f"]({REPORTS.as_posix()}/figures/")))
```

## Feature sets now available for Milestone 5

```python
rows = []
for s in ["primary", "strict_pretx", "posttx_ablation", "selected", "pca_variant"]:
    p = PROCESSED / f"features_{s}.json"
    if p.exists():
        j = json.loads(p.read_text())
        rows.append({"set": s, "n_columns": len(j["features"]), "batch_only": len(j.get("batch_only", [])), "aggregates": j.get("aggregates_included")})
display(pd.DataFrame(rows))
```
